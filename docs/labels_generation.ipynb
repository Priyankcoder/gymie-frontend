# Generate labels.json for Food-101

This notebook generates a **single source of truth** `labels.json`
using TensorFlow Datasets metadata.

- Dataset: **food101**
- Output: **labels.json** (101 labels, correct order)

Run all cells and download the output file from Kaggle.

```python
import tensorflow_datasets as tfds
import json
import os
```

```python
# Load dataset metadata only (no images downloaded)
builder = tfds.builder("food101")
info = builder.info

class_names = info.features["label"].names

print(f"✅ Number of classes: {len(class_names)}")
print("First 10 labels:", class_names[:10])
```

```python
# Save labels.json
LABELS_PATH = "labels.json"

with open(LABELS_PATH, "w") as f:
    json.dump(class_names, f, indent=2)

print(f"✅ labels.json saved at: {os.path.abspath(LABELS_PATH)}")
```

```python
# Verification (critical)
with open(LABELS_PATH) as f:
    loaded = json.load(f)

assert loaded == class_names
print("✅ Label order verified. File is safe to use.")
```
